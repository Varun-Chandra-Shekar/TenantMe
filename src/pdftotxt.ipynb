```python
#Libraries required
import pdfplumber
from pathlib import Path
import re
```

```python
#Converting PDF to raw text
PDF_PATH = Path("/Users/varunchandrashekar/Tenantmate/code/data/rawdata/act-2010-042.pdf")
OUT_PATH = Path("/Users/varunchandrashekar/Tenantmate/code/data/processed/act_raw.txt")


# make sure the folder exists
OUT_PATH.parent.mkdir(parents=True, exist_ok=True)

pages = []
with pdfplumber.open(PDF_PATH) as pdf:
    for page in pdf.pages:
        pages.append(page.extract_text() or "")   # "" guards against blank pages

# Join pages with the form-feed char "\f" so you can still see page breaks later
full_text = "\f".join(pages)
OUT_PATH.write_text(full_text, encoding="utf-8")

print(f"Extracted {len(pages)} pages, {len(full_text):,} characters")
```

```python
#Checking if the data is extracted correctly
text = Path("/Users/varunchandrashekar/Tenantmate/code/data/processed/act_raw.txt").read_text(encoding="utf-8")
print(repr(text[:500]))
```

```python
doubled = [ln for ln in text.split("\n") if re.search(r"([A-Za-z])\1{2,}", ln)]
print(f"Lines with tripled+ letters: {len(doubled)}")
for ln in doubled[:20]:
    print(repr(ln))
```
